# Step1: Blink the LED 

```python
# Step1: Blink the LED Light 

# blink.py
from machine import Pin
import time

led = Pin(2, Pin.OUT)  # Adjust the pin number as per your board's schematic

while True:
    led.on()
    time.sleep(1)
    led.off()
    time.sleep(1)
```

# Step2: Utilize Sensors and Peripherals 

```python
# Utilize all Sensors and Peripherals

from machine import Pin, I2C
import time

i2c = I2C(0, scl=Pin(22), sda=Pin(21))  # Adjust pins as necessary

# Example: Reading data from an I2C sensor
i2c_addr = 0x68  # Example I2C address
while True:
    data = i2c.readfrom(i2c_addr, 2)  # Reading 2 bytes from the sensor
    print(data)
    time.sleep(1)
```

# Read Data from the IMU Sensor

```python
from machine import I2C, Pin
import time

i2c = I2C(1, scl=Pin(22), sda=Pin(21))  # Initialize I2C with SCL on GPIO 22 and SDA on GPIO 21

imu_addr = 0x68  # Example I2C address for the IMU

while True:
    # Read sensor data, assuming the register is at 0x3B
    accel_data = i2c.readfrom_mem(imu_addr, 0x3B, 6)
    print("Accelerometer Data:", accel_data)
    time.sleep(1)
```

# Use the Onboard Microphone 

```python
import board
import audiobusio

mic = audiobusio.PDMIn(clock_pin=board.MIC_CLOCK, data_pin=board.MIC_DATA)

while True:
    samples = mic.record(160)  # Record a short audio sample
    print("Audio Sample:", samples)
```
